## Exercice 1 : premières requêtes, fichier

Afin de récupérer un lien vers le calendrier, je me suis connecté sur l'ENT de Rennes, j'ai seléctionné ma formation et ai généré un lien vers celui-ci.

Pour ce premier exercice, je me suis aidé du blog sur le site https://datascientist.fr qui explique comment utiliser la bibliothèque requests https://datascientist.fr/blog/tutoriel-python-telecharger-des-fichiers-depuis-des-url et comment manipuler des fichiers https://datascientist.fr/blog/la-manipulation-de-fichiers-en-python-tutoriel-complet

```python
import requests # Nous importons la bibliothèque `requests`

url = "https://planning.univ-rennes.fr/jsp/custom/modules/plannings/1348pR3k.shu" # Nous déclarons notre lien vers le calendrier dans une variable nommée `url`

reponse = requests.get(url) # Nous utilisons `requests.get()` pour récupérer le contenu de l'URL dans la variable `reponse`

with open("planning.ics", "w") as fichier: # Nous utilisons ensuite with open qui permet d'ouvrir notre fichier et de le refermer automatiquement à la fin du bloc afin de réduir le risque de fuite de ressources. Nous définissons le nom du nouveau fichier que nous voulons créer et utilisons le paramètre `w` afin d'être en mode écriture
    fichier.write(reponse.text) # La fonction .write permet d'écrire tout le texte qui a été récupéré dans la variable `reponse` dans le nouveau fichier planning
print("Le calendrier a été téléchargé et sauvegardé dans 'planning.ics'.") # Pour terminer, s'il n'y a aucune erreur, la phrase s'affichera et le fichier `planning.ics` sera créé.
```
