### Format d'URL pour l'API

Le site api.zippopotam.us indique le format pour appel à l'API : `https://api.zippopotam.us/country/city_code` où l'on doit remplacer 'country' par un code ('fr' ou 'us' pour France ou Etats-Unis par exemple) et 'city_code' par le code postal en France par exemple.

Sur le navigateur, nous avons atteint l'URL : `https://api.zippopotam.us/fr/35400` et la réponse affichée est :

```
{"post code": "35400", "country": "France", "country abbreviation": "FR", "places": [{"place name": "Saint-Malo", "longitude": "-2.0167", "state": "Bretagne", "state abbreviation": "A2", "latitude": "48.65"}]}
```

Nous supposons qu'il s'agit d'un format JSON.

### Bibliothèque pour http

On va utiliser la bibliothèque `requests` pour l'appel à l'API en ligne. (Pour l'installer `pip install requests`).

La doc de requests est très simple, le lien [quickstart](https://requests.readthedocs.io/en/latest/user/quickstart/) suffit pour ce que nous voulons faire :

Plus particulièrement la partie : [https://requests.readthedocs.io/en/latest/user/quickstart/#json-response-content](https://requests.readthedocs.io/en/latest/user/quickstart/#json-response-content)


Un exemple sur la doc nous permet de savoir comment atteindre une URL qui renvoie un JSON :

```python
r = requests.get('https://api.github.com/events')
r.json()
```

### Code proposé :

Puisque le décodage d'un format JSON renvoie un dictionnaire nous pouvons écrire le code suivant :

```python
import requests as rq

def get_coor(city_code:str, country:str='fr') -> tuple:
    """ Renvoie le nom et les coordonnées géographiques d'une ville à partir de son code postal et de son pays """
    # utilisation d'une f-string pour insérer les variables dans l'url
    url = f'https://api.zippopotam.us/{country}/{city_code}'
    # on récupère les données de l'API
    response = rq.get(url)
    # on les transforme en dictionnaire
    data = response.json()
    # si le code postal n'existe pas ou ne renvoie pas de bonne réponse, on renvoie None
    if 'places' not in data:
        return None, (None, None)
    return data['places'][0]['place name'], (float(data['places'][0]['latitude']), float(data['places'][0]['longitude']))
```
