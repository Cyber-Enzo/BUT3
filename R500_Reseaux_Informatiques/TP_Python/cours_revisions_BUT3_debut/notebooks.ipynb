```python
import datetime

jour = int(input("Jour de naissance"))
mois = int(input("Mois de naissance : "))
annee = int(input("Année de naissance : "))
date = datetime.date.today()

print(f"Vous êtes né le {jour}/{mois}/{annee}")
print(date)

def is_major(jour,mois,annee)-> bool:
    annee_actuelle = date.year
    mois_actuel = date.month
    jour_actuel = date.day
    
    # 3. Calculer l'âge de base
    age = annee_actuelle - annee
    
    # 4. Vérifier si l'anniversaire n'est PAS encore passé cette année
    # Si la date actuelle (mois, jour) est avant (mois, jour) de naissance :
    if (mois_actuel, jour_actuel) < (mois, jour):
        age = age - 1  # ou age -= 1
        
    return age >= 18
resultat = is_major(jour,mois,annee)
print(resultat)
```
