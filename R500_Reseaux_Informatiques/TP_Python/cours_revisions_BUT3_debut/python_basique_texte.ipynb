````python
# 2.4 : créer une fonction nommée "codage(txt)" qui prend un texte en paramètre
# et appèle la fonction du 2.3 sur chacune des lettres de txt en décalant d'abord de 1 puis de 2 caractères,
# puis de 3 etc. pour chaque lettre du message.
# la fonction renvoie le résultat de l'encodage.
# exemple : codage("bonjour") renvoie "cqqntay"


def codage(txt):
    txt_min = txt.lower()
    resultat = ""
    decalage = 1

    for lettre in txt_min:
        lettre_codee = cesar(lettre, decalage)
        resultat = resultat + lettre_codee
        decalage = decalage +1
    return resultat 



try:
    import corrige
    corrige.test_import()
    corrige.codage(codage)
except Exception as err:
    print("nok ...", err)
````

# Exo 5 : map, filter, sorted

Les fonctions `map`, `filter` et `sorted` sont des fonctions intégrées en Python qui permettent de manipuler des listes (ou d'autres itérables) en utilisant des fonctions lambda (fonctions anonymes) pour effectuer des opérations spécifiques. Voici une introduction avec des exemples pour chacune de ces fonctions :

1. **`map`** :

La fonction `map` est utilisée pour appliquer une fonction donnée à chaque élément d'une liste (ou d'un autre iterable) et renvoie un nouvel iterable contenant les résultats de l'application de cette fonction à chaque élément. Voici un exemple avec une fonction lambda pour doubler chaque élément d'une liste de nombres :

```python
numbers = [1, 2, 3, 4, 5]
doubled_numbers = list(map(lambda x: x * 2, numbers)) 
# map renvoie un iterable donc il faut faire list(map(...)) pour obtenir une liste
print(doubled_numbers)  # Résultat : [2, 4, 6, 8, 10]
```

2. **`filter`** :

La fonction `filter` permet de filtrer les éléments d'une liste en fonction d'une condition spécifiée dans une fonction lambda. Elle renvoie un nouvel iterable contenant uniquement les éléments pour lesquels la condition est vraie. Voici un exemple pour filtrer les nombres pairs d'une liste :

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
# filter renvoie un iterable donc il faut faire list(filter(...)) pour obtenir une liste
print(even_numbers)  # Résultat : [2, 4, 6, 8]
```

3. **`sorted`** :

La fonction `sorted` est utilisée pour trier les éléments d'une liste (ou d'un iterable) selon un critère spécifié dans une fonction lambda. Elle renvoie une nouvelle liste contenant les éléments triés. Voici un exemple pour trier une liste de mots par leur longueur :

```python
words = ["chien", "chat", "éléphant", "souris"]
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)  # Résultat : ['chat', 'chien', 'souris', 'éléphant']
```

4. **`reduce`** :

La fonction `reduce` est une fonction de la bibliothèque `functools` de Python (à partir de Python 3, elle est incluse dans le module `functools`). Elle est utilisée pour effectuer une opération cumulative sur une séquence d'éléments en utilisant une fonction lambda.

La fonction `reduce` prend une fonction lambda (fonction anonyme) qui prend deux arguments, ainsi qu'une séquence d'éléments. Elle applique cette fonction de manière cumulative à la séquence, en prenant le résultat de l'opération précédente et l'élément suivant de la séquence comme arguments pour la prochaine opération. Elle renvoie finalement la valeur accumulée.

Voici un exemple où nous utilisons `reduce` pour calculer la somme des éléments d'une liste :

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
sum_result = reduce(lambda x, y: x + y, numbers)
print(sum_result)  # Résultat : 15
```

Dans cet exemple, la fonction lambda prend deux arguments `x` et `y`, et elle additionne ces deux valeurs à chaque étape de la réduction. Ainsi, `reduce` accumule les éléments de la liste pour calculer la somme totale.

La fonction `reduce` est particulièrement utile lorsque vous devez effectuer des opérations itératives sur une séquence, en maintenant un état cumulatif à chaque étape. Elle offre une approche élégante pour résoudre ce genre de problèmes en Python.
